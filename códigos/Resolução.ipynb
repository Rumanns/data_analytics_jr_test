```python
import sqlite3
import pandas as pd
import os

# Caminhos para as pastas contendo os arquivos CSV
pasta_escolas = 'Data/Data/Escolas'
pasta_perfil = 'Data/Data/Perfil dos educandos'

# Caminho para o banco de dados
caminho_banco_dados = 'test_analytics.db'

# Verifica se o banco de dados existe e o deleta, se necessário
if os.path.exists(caminho_banco_dados):
    print(f"Banco de dados '{caminho_banco_dados}' já existe. Deletando...")
    os.remove(caminho_banco_dados)
    print(f"Banco de dados '{caminho_banco_dados}' deletado com sucesso.")

# Conecta ao banco de dados SQLite (ou cria um se não existir)
conn = sqlite3.connect(caminho_banco_dados)
cursor = conn.cursor()

# Função para carregar arquivos CSV de uma pasta específica
def carregar_csv_para_sqlite(pasta):
    # Lista todos os arquivos CSV na pasta
    arquivos_csv = [f for f in os.listdir(pasta) if f.endswith('.csv')]

    # Para cada arquivo CSV, criar uma tabela e carregar os dados
    for arquivo in arquivos_csv:
        # Nome da tabela é o mesmo nome do arquivo (sem a extensão .csv)
        nome_tabela = os.path.splitext(arquivo)[0]
        
        # Lê os arquivos CSV
        df = pd.read_csv(os.path.join(pasta, arquivo), delimiter=";", encoding="latin1")
#        print(len(df.columns), df.columns)
        
        # Cria a tabela no SQLite
        df.to_sql(nome_tabela, conn, if_exists='replace', index=False)
        
        print(f'Tabela {nome_tabela} criada com sucesso')

# Carrega os arquivos da pasta "Escolas"
print("Carregando dados da pasta 'Escolas'...")
carregar_csv_para_sqlite(pasta_escolas)

# Carrega os arquivos da pasta "Perfil"
print("\nCarregando dados da pasta 'Perfil dos educandos'...")
carregar_csv_para_sqlite(pasta_perfil)

# Função para listar as tabelas criadas no banco de dados
def listar_tabelas(cursor):
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tabelas = cursor.fetchall()
    print("\nTabelas criadas no banco de dados:")
    for tabela in tabelas:
        print(tabela[0])

# Lista as tabelas criadas
listar_tabelas(cursor)

# Fecha a conexão com o banco de dados
conn.close()
```

```python
import pandas as pd
import os
import seaborn as sns

path_escolas = "Data/Data/Escolas"
path_alunos = "Data/Data/Perfil dos educandos"

def analises(pasta):
    #Define os padrões de coluna de cada tipo de arquivo
    if pasta == "Data/Data/Escolas":
        colunas = ['DRE', 'CODESC', 'TIPOESC', 'CEU', 'DIRETORIA', 'SUBPREF', 'ENDERECO',
    'NUMERO', 'BAIRRO', 'CEP', 'TEL1', 'TEL2', 'FAX', 'SITUACAO', 'CODDIST',
    'DISTRITO', 'SETOR', 'CODINEP', 'EH', 'DT_CRIACAO', 'ATO_CRIACAO',
    'DT_AUTORIZA', 'NOME_ANT', 'LATITUDE', 'LONGITUDE', 'DATABASE']
        
    elif pasta == "Data/Data/Perfil dos educandos":
        colunas = ['DRE', 'CODESC', 'TIPOESC', 'NOMESC', 'DISTRITO', 'SETOR', 'ANO',
       'REDE', 'MODAL', 'DESCSERIE', 'PERIODO', 'TURNO', 'DESCTURNO', 'SEXO',
       'IDADE', 'NEE', 'RACA', 'QTDE', 'DATABASE']

    #Lista os arquivos na pasta
    arquivos_csv = [f for f in os.listdir(pasta) if f.endswith('.csv')]
        
    #Cria uma lista de DataFrames para juntar eles depois
    dfs = []
    for arquivo in arquivos_csv:
        # Lê os arquivos CSV
        df = pd.read_csv(os.path.join(pasta, arquivo), delimiter=";", encoding="latin1")
        
        #Coloca todos os títulos de colunas em maiúsculo para facilitar a junção das colunas
        df.columns = df.columns.str.upper()
        
        #Deleta colunas que estejam fora das colunas estabelecidas
        for coluna in df.columns:
            if coluna not in colunas:
                df.drop(coluna, axis=1, inplace=True)
                
        dfs.append(df)

        print(arquivo, '\n', len(df.columns), df.columns)
    
    
#    print(dfs)
    
# Carrega os arquivos da pasta "Perfil"
print("\nCarregando dados da pasta 'Perfil dos educandos'...")
analises(path_escolas)

# Carrega os arquivos da pasta "Perfil"
print("\nCarregando dados da pasta 'Perfil dos educandos'...")
analises(path_alunos)
```
